```python
df[(df["Category"] == "Electronics") & (df["Country"] == "USA")]                  # fetching data  using logic operator (and)
```

```python
df[df["Quantity"] > 10]                  # fetching data using comparison operator
```

```python
df[df["CustomerName"].str.startswith('A')]               # filters with starting letter in a specific feature [column]
```

```python
df[df["CustomerName"].str.endswith('a')]               # filters with ending letter in a specific feature [column]
```

```python
df[df["Country"].isin(["UAE", "Australia", "Spain"])]               # takes full row that are in the list of countries
```

```python
df[~df["Country"].isin(["UAE", "Australia", "Spain"])]               # '~' is like NOT in the list
```

```python
# now if I want to update a column with a Customer name, what can I do?

df.loc[df["CustomerName"] == "Stephen Kennedy"]                   # so first I see whether the customername is unique or any repeatation happen
```

```python
df.loc[df["CustomerName"] == "Stephen Kennedy", 'Product'] = "MacBook"

# > df.loc[df["CustomerName"] == "Stephen Kennedy", < for locating the data; >, 'Product'< is the column to update; 
# and the >] = "MacBook"< is the value to update
```

```python
# now if i check again his product
df.loc[df["CustomerName"] == "Stephen Kennedy"]                   # it will be updated
```

```python
# another example

df.loc[df["Country"] == "USA"]
```

```python
df.loc[df["Country"] == "USA", "Country"] = "United States"              # now I change the USA --> United States
```

```python
df.loc[df["Country"] == "United States"]                     # now if i run with United States, there won't be any USA as country
```

```python
# to make every country with Caps
df["Country"] = df["Country"].str.upper()
```
